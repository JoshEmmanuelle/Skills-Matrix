# Exploring Titanic dataset with Pandas

The titanic dataset is a classic dataset used for data analysis and machine learning. It contains information about the passengers on the Titanic, including

```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/MicrosoftDocs/mslearn-introduction-to-machine-learning/main/Data/titanic.csv"
```

```python
titanic = pd.read_csv(url)
```

```python
titanic
```

```python
# First, let's inspect the data quality
print("Dataset shape:", titanic.shape)
print("\nData types and missing values:")
print(titanic.info())
print("\nMissing values:")
print(titanic.isnull().sum())
print("\nBasic statistics:")
print(titanic.describe())
```

```python
# Data cleaning steps
titanic_clean = titanic.copy()

# 1. Remove rows with missing critical values (PassengerId should be unique)
# Drop rows where Survived is missing (the target variable)
titanic_clean = titanic_clean.dropna(subset=['Survived'])

# 2. Handle Age column (most missing values)
# Fill Age with median age by passenger class
titanic_clean['Age'] = titanic_clean.groupby('Pclass')['Age'].transform(
    lambda x: x.fillna(x.median())
)

# 3. Handle Embarked column (few missing values)
# Fill with most common port
titanic_clean['Embarked'] = titanic_clean['Embarked'].fillna(titanic_clean['Embarked'].mode()[0])

# 4. Drop Cabin column (too many missing values ~77%)
titanic_clean = titanic_clean.drop('Cabin', axis=1)

# 5. Convert categorical variables to numeric
titanic_clean['Sex'] = titanic_clean['Sex'].map({'male': 1, 'female': 0})
titanic_clean['Embarked'] = titanic_clean['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})

# 6. Drop PassengerId and Name (not useful for analysis)
titanic_clean = titanic_clean.drop(['PassengerId', 'Name', 'Ticket'], axis=1)

print("Cleaned dataset shape:", titanic_clean.shape)
print("\nRemaining missing values:")
print(titanic_clean.isnull().sum())
print("\nCleaned data:")
print(titanic_clean.head())
```
